```python
import ultralytics
ultralytics.checks()
```

## Train the model

```python
model = ultralytics.YOLO("YOLOv8m.yaml")

dataset = "data/scenes/scenes.yaml"

project = "model/tichu_yolov8m"
experiment = "tichu_yolov8m"

batch_size = 32

results = model.train(data=dataset, epochs=50, project=project, name=experiment, batch=batch_size, device=0, patience=5, imgsz=640, verbose=True, val=True)
```

```python
results = model.val(data=dataset, split="test", imgsz=640, device=0)
```
