```python
import ultralytics
ultralytics.checks()
```

## Train the model

```python
model = ultralytics.YOLO("YOLOv8l.yaml")

dataset = "data/scenes/scenes.yaml"

project = "model/"
experiment = "tichu_yolov8l"

batch_size = 64

results = model.train(data=dataset, epochs=200, project=project, name=experiment, batch=batch_size, device=0, patience=5, imgsz=640, verbose=True, val=True)
```

## Test the model

```python
model_path = "model/tichu_yolov8l/weights/best.pt"
model = ultralytics.YOLO(model_path)
results = model.val(data=dataset, split="test", imgsz=640, device=0)
```

```python
import cv2
from ultralytics import YOLO

# Load the trained model (adjust the path as needed)
model = YOLO("model/tichu_yolov8l/weights/best.pt")

# Define the video source (0 for webcam, or provide a video file path)
video_source = 0
cap = cv2.VideoCapture(video_source)

# Get frame dimensions and frames per second from the capture device
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
# If fps is not available, set a default value (e.g., 20)
if fps == 0:
    fps = 20

# Define the codec and create a VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*'XVID')
output_file = "output.avi"
out = cv2.VideoWriter(output_file, fourcc, fps, (width, height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Run the model on the current frame (non-streaming mode)
    results = model.predict(source=frame, stream=False)
    
    # Annotate the frame (assuming results contains one result)
    annotated_frame = results[0].plot()

    # Write the annotated frame to the output video file
    out.write(annotated_frame)

    # Optionally, display the frame in a window
    cv2.imshow("YOLOv8 Video Stream", annotated_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()
```
